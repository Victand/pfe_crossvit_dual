```python
import pandas as pd
import json

# Le nom de ton fichier Excel exporté par Numbers
NOM_FICHIER_EXCEL = "H:\Documents\Ing 3\PFE\data\Data_Projet\Data_v2_excel.xlsx"

# 1. Lecture du fichier Excel (simple et fiable)
# Si ton fichier a plusieurs feuilles, tu peux spécifier le nom de la feuille
# ou 0 pour la première feuille.
try:
    df = pd.read_excel(NOM_FICHIER_EXCEL, sheet_name=0)

    # 2. Conversion en JSON
    # 'orient='records'' donne une liste de dictionnaires (le format le plus couramment utilisé en JSON).
    # 'force_ascii=False' permet de garantir la conservation des accents et caractères spéciaux.
    json_data = df.to_json(orient="records", indent=4, force_ascii=False)

    # 3. Sauvegarder dans un fichier .json en UTF-8
    with open(
        "H:\Documents\Ing 3\PFE\data\Data_Projet\data.json", "w", encoding="utf-8"
    ) as f:
        f.write(json_data)

    print("Grrr ! Conversion XLSX vers JSON réussie et propre !")

except FileNotFoundError:
    print(f"Erreur : Le fichier {NOM_FICHIER_EXCEL} n'a pas été trouvé.")
except Exception as e:
    print(f"Une erreur est survenue lors de la conversion : {e}")
```

```python
import os

# --- Chemins des fichiers (Lecture et Écriture) ---

# Le fichier en entrée/sortie unique
NOM_FICHIER_JSON = r"H:\Documents\Ing 3\PFE\data\Data_Projet\data.json"

# --- 1. CHARGEMENT DES DONNÉES BRUTES ---
donnees_json_brutes = None
try:
    if not os.path.exists(NOM_FICHIER_JSON):
        raise FileNotFoundError(f"Le fichier JSON n'existe pas : {NOM_FICHIER_JSON}")

    with open(NOM_FICHIER_JSON, "r", encoding="utf-8") as f:
        print(f"✅ Lecture du fichier en cours : {NOM_FICHIER_JSON}")
        donnees_json_brutes = json.load(f)

except FileNotFoundError as e:
    print(f"\nGrrr ! ERREUR : {e}")
    exit()  # Arrête le script si le fichier n'est pas trouvé
except json.JSONDecodeError:
    print(f"\nGrrr ! ERREUR : Le fichier {NOM_FICHIER_JSON} n'est pas un JSON valide.")
    exit()

# --- 2. TRAITEMENT ET NETTOYAGE (grrr) ---

# On s'assure que les données sont une liste non vide
if not isinstance(donnees_json_brutes, list) or len(donnees_json_brutes) < 2:
    print(
        "\nGrrr ! ERREUR : Le fichier ne contient pas une liste de données exploitable."
    )
    exit()

# On ignore la première ligne (indice 0) qui contient les en-têtes dupliqués
donnees_utiles = donnees_json_brutes[1:]

donnees_finales = []

for item in donnees_utiles:
    valeur_epine = item.get("Unnamed: 1")
    valeur_epine_finale = 0  # Valeur par défaut

    # Conversion de 'epine' en ENTIER (int)
    try:
        # Assure-toi que la valeur est traitée comme une chaîne pour remplacer la virgule,
        # puis convertie en float pour gérer les '.0', puis en int pour le format final.
        valeur_flottante = float(str(valeur_epine).replace(",", "."))
        valeur_epine_finale = int(valeur_flottante)

    except Exception:
        # En cas d'erreur de conversion (valeur nulle ou non numérique), on utilise la valeur par défaut
        valeur_epine_finale = 0

    nouvel_objet = {
        "id": item.get("Data_mazene_v2"),  # L'ID est laissé tel quel (chaîne ou int)
        "epine": valeur_epine_finale,  # Le nouvel entier
    }

    donnees_finales.append(nouvel_objet)


# --- 3. SAUVEGARDE DU RÉSULTAT FINAL ---

# Écrasement du fichier d'origine avec les données propres
try:
    with open(
        "H:\Documents\Ing 3\PFE\data\Data_Projet\data_presque_propre.json",
        "w",
        encoding="utf-8",
    ) as f:
        json.dump(donnees_finales, f, indent=4)

    print(
        f"\n✅ Nettoyage et conversion terminés. Le fichier a été mis à jour : {NOM_FICHIER_JSON}"
    )

except Exception as e:
    print(f"\nGrrr ! Échec de la sauvegarde. Erreur : {e}")
```
